```python
from sklearn.preprocessing import StandardScaler

train_x = df1[['ASC-16 amount(5-40mg)', 'Temperature(200-280゜C)',
              'Cs/Pb(0.6-1.8)', 'Yb/Pb(1.5-4)', 'OA amount(0-2ml)']].to_numpy()
xscaler = StandardScaler()
train_x_transformed = xscaler.fit_transform(train_x)

# i flip it around since skopt takes minimization
train_y = -df1['PLQY(%)'].to_numpy()
yscaler = StandardScaler()
train_y_transformed = yscaler.fit_transform(train_y.reshape(-1, 1))
```

```python
from sklearn.ensemble import RandomForestRegressor 
from sklearn.model_selection import LeaveOneOut

# Initialize array to store feature importances
feature_importances = np.zeros((len(df), 5))

loo = LeaveOneOut()

# Perform LOOCV and collect feature importances for each iteration
# 23 samples, 1 sample remove

for train_index, test_index in loo.split(train_x_transformed):
    X_train, X_test = train_x_transformed[train_index], train_x_transformed[test_index]
    y_train, y_test = train_y_transformed[train_index], train_y_transformed[test_index]

    # Clone the model to ensure it's a fresh model for each iteration
    model = RandomForestRegressor()
    model.fit(X_train, y_train.ravel())

    # Store the feature importances
    feature_importances[test_index] = model.feature_importances_

# Calculate average feature importances over all LOOCV iterations
for i, variable in enumerate(['ASC-16 amount(5-40mg)', 'Temperature(200-280゜C)',
                              'Cs/Pb(0.6-1.8)', 'Yb/Pb(1.5-4)', 'OA amount(0-2ml)']):
    print(f"Feature {variable}: {feature_importances[:,i].mean():.5} +/- {feature_importances[:,i].std():.5}")
```

```python
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.metrics import mean_squared_error, r2_score

loo = LeaveOneOut()

# Function to perform LOOCV and return true vs predicted values for the full dataset
def loocv_full_dataset(regressor, X, y):
    true, predicted = [], []
    mse_scores, r2_scores = [], []

    for train_index, test_index in loo.split(X):
        X_train, X_test = X[train_index], X[test_index]
        y_train, y_test = y[train_index], y[test_index]

        # Training on all data except one
        regressor.fit(X_train, y_train.ravel())
        # Predicting on the full dataset (including the one left out)
        y_pred_full = regressor.predict(X)

        true.extend(y.ravel())
        predicted.extend(y_pred_full)

        mse_scores.append(mean_squared_error(y, y_pred_full))
        r2_scores.append(r2_score(y, y_pred_full))

    return true, predicted, np.mean(mse_scores), np.std(mse_scores), np.mean(r2_scores), np.std(r2_scores)

#List of regressors to be compared
regressors = {
    "Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(),
    "Random Forest": RandomForestRegressor(),
    "Gradient Boosting": GradientBoostingRegressor(),
    "Gaussian Process": GaussianProcessRegressor()
}



for i, (name, regressor) in enumerate(regressors.items()):
    true, predicted, mse_mean, mse_std, r2_mean, r2_std = loocv_full_dataset(regressor, train_x_transformed, train_y_transformed)
    
    # Plot for each regressor
    fig, ax = plt.subplots(1, 1, figsize=(5, 3))

    ax.scatter(true, predicted, s=5)
    ax.axline((1, 1), slope=1)
    ax.set_title(f'{name} (MSE: {mse_mean:.3f}±{mse_std:.3f}, R2: {r2_mean:.3f}±{r2_std:.3f})')
    ax.set_xlabel('True Values')
    ax.set_ylabel('Predicted Values')

plt.tight_layout()
plt.show()
```

# Optimization Part

```python
from skopt import Optimizer

lowerbounds = np.array([5,200,0.6,1.5,0])
upperbounds = np.array([40,260,1.8,4,2])

lowerbounds = xscaler.transform(lowerbounds.reshape(1, -1)).tolist()[0]
upperbounds = xscaler.transform(upperbounds.reshape(1, -1)).tolist()[0]

opt = Optimizer([(lower, upper) for lower, upper in zip(lowerbounds, upperbounds)],
                   base_estimator='GP', n_initial_points=0,
                   acq_func="EI", acq_optimizer="lbfgs", random_state=0)

# remember here that
opt.tell(train_x_transformed.tolist(), train_y_transformed.ravel().tolist(), fit=True)
new_x = opt.ask(n_points=6)
new_x_untransformed = xscaler.inverse_transform(new_x)
pd.DataFrame(new_x_untransformed, columns=['ASC-16 amount(5-40mg)', 'Temperature(200-280゜C)',
                                       'Cs/Pb(0.6-1.8)', 'Yb/Pb(1.5-4)', 'OA amount(0-2ml)'])
```
